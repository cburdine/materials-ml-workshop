```python
xrd_df = pd.read_csv('xrd_dataset_clean.csv')

# Generate a list of the crystal systems in the dataset:
symm_symbols = sorted(set(xrd_df['symmetry_symbol']))

symm_symbols_canonical = [ SpaceGroup(s).symbol for s in symm_symbols ]
```

```python
from pymatgen.symmetry.groups import SpaceGroup

sg_symbols = []
for n in range(1,231):
    sg = SpaceGroup.from_int_number(n)
    sg_symbols.append(sg.symbol)
```

```python
from tqdm import tqdm

sg_data = {}
with MPRester() as mpr:

    for symbol in tqdm(sg_symbols):
        docs = mpr.materials.summary.search(
            energy_above_hull=(0, 1.0),
            spacegroup_symbol=symbol,
            fields=['material_id', 'formula_pretty', 'energy_above_hull'],
        )

        if docs:
            sg_data[symbol] = [dict(doc) for doc in docs]
```

```python
from pymatgen.analysis.diffraction.xrd import XRDCalculator
from pymatgen.symmetry.analyzer import SpacegroupAnalyzer


df_data = {
    'mp_id' : [],
    'formula': [],
    'composition': [],
    'crystal_system': [],
    'symmetry_symbol': [],
    'cell_params': [],
    'energy_above_hull': [],
    'xray_peaks': [],
    'xray_intensities': []
}
with MPRester(mute_progress_bars=True) as mpr:
    for i, (sg, doc_list) in enumerate(sg_data.items()):
        print(i, sg)
        
        for doc in tqdm(doc_list[:min(4096, len(doc_list))]):
            mpid = doc['material_id']
            if mpid in seen_mpids:
                continue
            
            structure = mpr.get_structure_by_material_id(mpid)
 
            sga = SpacegroupAnalyzer(structure)
            conventional_structure = sga.get_conventional_standard_structure()
            sga2 = SpacegroupAnalyzer(conventional_structure)

            
            conventional_sg = sga2.get_space_group_symbol()
            conventional_system = sga2.get_crystal_system()

            composition = str(get_composition(conventional_structure))
            cell_params = str(get_cell_params(conventional_structure))
            
            # this example shows how to obtain an XRD diffraction pattern
            # these patterns are calculated on-the-fly from the structure
            calculator = XRDCalculator(wavelength="CuKa")
            pattern = calculator.get_pattern(conventional_structure)
            peaks = '['+','.join([f'{s:.3f}' for s in pattern.x])+']'
            intensities = '['+','.join([f'{s:.3f}' for s in pattern.y])+']'

            df_data['mp_id'].append(mpid)
            df_data['formula'].append(doc['formula_pretty'])
            df_data['composition'].append(composition)
            df_data['crystal_system'].append(conventional_system)
            df_data['symmetry_symbol'].append(conventional_sg)
            df_data['cell_params'].append(cell_params)
            df_data['energy_above_hull'].append(doc['energy_above_hull'])
            df_data['xray_peaks'].append(peaks)
            df_data['xray_intensities'].append(intensities)
```

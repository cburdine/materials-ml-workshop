```python
import numpy as np
import sklearn
import matplotlib.pyplot as plt
```

```python
x_data = np.random.uniform(-30, 30, size=100)
y_data = 10 - 0.01*x_data**2 + np.random.normal(0,2,size=100)
```

```python
plt.figure()
plt.scatter(x_data, y_data)
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_nontrain, y_train, y_nontrain = train_test_split(
    x_data,
    y_data,
    test_size=0.2
)
```

```python
x_validation, x_test, y_validation, y_test = \
    train_test_split(
        x_nontrain,
        y_nontrain,
        test_size=0.5
    )
```

```python
# compute mean and std. dev. of training data
mu_x = np.mean(x_train)
sigma_x = np.std(x_train)

z_train = (x_train - mu_x)/sigma_x
z_validation = (x_validation - mu_x)/sigma_x
z_test = (x_test - mu_x)/sigma_x

mu_y = np.mean(y_train)
sigma_y = np.std(y_train)

# standardize the labels (optional):
u_train = (y_train - mu_y)/sigma_y
u_validation = (y_validation - mu_y)/sigma_y
u_test = (y_test - mu_y)/sigma_y
```

```python
plt.figure()
plt.scatter(z_train, u_train, label='Train Set')
plt.scatter(z_validation, u_validation, label='Val Set')
plt.scatter(z_test, u_test, label='Test Set')
plt.xlabel('z')
plt.ylabel('u')
plt.grid()
plt.legend()
plt.show()
```

```python
# fit polynomial model to standardized data:
poly_model_1 = np.poly1d(np.polyfit(z_train, u_train, deg=1))
poly_model_2 = np.poly1d(np.polyfit(z_train, u_train, deg=2))
poly_model_3 = np.poly1d(np.polyfit(z_train, u_train, deg=3))


# evaluate polynomial model at (standardized) point z:
z_eval = np.linspace(-2, 2, 1000)
u_hat_1 = poly_model_1(z_eval)
u_hat_2 = poly_model_2(z_eval)
u_hat_3 = poly_model_3(z_eval)


plt.figure()
plt.plot(z_eval, u_hat_1, label='Degree 1 fit')
plt.plot(z_eval, u_hat_2, label='Degree 2 fit')
plt.plot(z_eval, u_hat_3, label='Degree 3 fit')
plt.scatter(z_validation, u_validation)
plt.grid()
plt.xlabel('z')
plt.ylabel('u')
plt.legend()
plt.show()
```

```python
# un-standardize the curve predicted by our best fit
y_hat_2 = u_hat_2*sigma_y + mu_y
x_eval = z_eval*np.std(x_data) + np.mean(x_data)

plt.figure()
plt.plot(x_eval, y_hat_2)
plt.scatter(x_test, y_test)
plt.grid()
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

# Application: Classifying Perovskites

```python
import requests

CSV_URL = 'https://raw.githubusercontent.com/cburdine/materials-ml-workshop/main/MaterialsML/supervised_learning/perovskites.csv'

r = requests.get(CSV_URL)
with open('perovskites.csv', 'w') as f:
    f.write(r.text)
```

```python
import pandas as pd
```

```python
# load the cav file into a Pandas dataframe
perovskite_df = pd.read_csv('perovskites.csv')
```

```python
display(perovskite_df)
```

```python
import numpy as np

# extract the features into a numpy array:
a_electronegativity = np.array(perovskite_df['EN(A)'])
b_electronegativity = np.array(perovskite_df['EN(B)'])

goldschmidt_tolerance = np.array(perovskite_df['tG'])

# combine features into columns of a Nx3 numpy array:
features = np.array([
    a_electronegativity,
    b_electronegativity,
    goldschmidt_tolerance
]).T

```

```python
# dataset labels:
structures = np.array(perovskite_df['Lowest distortion'])

# reduce to distrinct values:
distinct_structures = list(set(structures))
print(distinct_structures)
```

```python
# determine how many of each lowest distortion phase are present:
bar_counts = np.array([
    len(structures[(structures == structure_type)])
    for structure_type in distinct_structures
])

plt.figure()
bar_locations = list(range(len(distinct_structures)))
plt.barh(bar_locations, bar_counts)
plt.gca().set_yticks(bar_locations)
plt.gca().set_yticklabels(distinct_structures)
plt.xlabel('Count')
plt.show()
```

```python
# determine the indices of unlabeled  data:
clean_data_idxs = (structures != '-')

# remove unlabeled data from features:
clean_features = features[clean_data_idxs]
clean_structures = structures[clean_data_idxs]

# assign binary classifier labels +1/-1 for cubic/non-cubic:
binary_classifier_labels = np.array([
    1 if struct == 'cubic' else -1
    for struct in structures
])

print('Features shape:', clean_features.shape)
print('Binary classifier labels shape:', binary_classifier_labels.shape) 
```

```python
# separate cubic vs. noncubic structures:
cubic_data = features[binary_classifier_labels == 1]
noncubic_data = features[binary_classifier_labels == -1]

plt.figure()
plt.scatter(cubic_data[:,0], cubic_data[:,1], s=0.8, label='Cubic')
plt.scatter(noncubic_data[:,0], noncubic_data[:,1], s=0.8, label='Non-Cubic')
plt.xlabel('EN(A)')
plt.ylabel('EN(B)')
plt.legend()
plt.show()
```

```python
hist_bins = np.linspace(
    np.min(goldschmidt_tolerance),
    np.max(goldschmidt_tolerance),
    50)

plt.figure()
plt.hist(cubic_data[:,2], bins=hist_bins, alpha=0.4, label='Cubic')
plt.hist(noncubic_data[:,2], bins=hist_bins, alpha=0.4, label='Non-Cubic')
```

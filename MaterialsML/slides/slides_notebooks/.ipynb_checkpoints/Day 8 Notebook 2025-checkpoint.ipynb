## Exercise: Applying PCA

```python
from scipy.stats import ortho_group, multivariate_normal
import numpy as np

# generate multivariate normal random dataset:
D = np.diag([1e-2]*20+list(np.random.uniform(1,10,10)))
U = ortho_group.rvs(D.shape[0])
mu = np.random.uniform(0,100, size=30)

data_x = multivariate_normal.rvs(mean=mu,cov=(U @ D @ U.T), size=4000)
```

```python
print(data_x.shape)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# standardize the data
scaler = StandardScaler()
scaler.fit(data_x)
data_z = scaler.transform(data_x)

# fit a full PCA to data to determine explained variances:
full_pca = PCA(n_components=30)
full_pca.fit(data_z)

# extract explained variances (entries of D):
variances = full_pca.explained_variance_
```

```python
# plot the variances of the principal components:
```

```python
variances.cumsum() / variances.sum()
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.title('Explained Variances for Principal Components')
plt.bar(np.arange(1,len(variances)+1), variances)
plt.show()
```

```python
partial_pca = PCA(n_components=10)
data_u = partial_pca.fit_transform(data_z)
```

```python
plt.figure()
plt.scatter(data_z[:,0], data_z[:,1])
plt.xlabel('First Principal component (U_1)')
plt.ylabel('Second Principal component (U_2)')
plt.show()
```

# Example: Estimating the density of states of Graphene

```python
import numpy as np

def graphene_energies(n = 100):
    rcell = np.array([[1, 0.57735], [0, 1.1547]])
    b = 2*np.pi*np.linspace(0,1,n)
    ux, uy = np.meshgrid(b,b)
    k = np.stack([ux.flatten(), uy.flatten()]).T
    k = (k @ rcell).T
    E = np.sqrt(1 + 4*np.cos(k[0]/2)**2 + 4*np.cos(k/2).prod(0))
    return np.concat((E,-E))

allowed_energies = graphene_energies()
```

```python
print(allowed_energies.shape)
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.hist(allowed_energies, bins=100)
plt.show()
```

```python
from scipy.stats import gaussian_kde
```

```python
kde = gaussian_kde(allowed_energies, bw_method=0.02)

E = np.linspace(-3, 3, 1000)
dos = kde(E)
```

```python
# plot our density of states
plt.figure()
plt.xlim((-3,3))
plt.fill_between(E, dos)
plt.yticks([])
plt.xlabel('Energy')
plt.ylabel('Density of States (arb. units)')
plt.show()
```

## Application: Classifying Superconductors

```python
import requests

CSV_URL = 'https://raw.githubusercontent.com/cburdine/materials-ml-workshop/main/MaterialsML/unsupervised_learning/matml_supercon_cleaned.csv'

r = requests.get(CSV_URL)
with open('matml_supercon_cleaned.csv', 'w', encoding='utf-8') as f:
    f.write(r.text)
```

```python
import pandas as pd
```

```python
data_df = pd.read_csv('matml_supercon_cleaned.csv')
```

```python
display(data_df)
```

```python

# remove any rows of data that have a corrupted composition:
data_df = data_df[ data_df['Composition'].isnull() == False ]

# remove any materials with unphysically high Tc:
data_df = data_df[ data_df['Tc (K)'] < 400 ]
```

```python
display(data_df)
```

```python
import ast

# generate a list elements contained in the dataset:


ELEMENTS = set()
for v in data_df['Composition'].values:
    ELEMENTS |= set(ast.literal_eval(v).keys())

ELEMENTS = list(ELEMENTS)
```

```python
print(ELEMENTS)
```

```python
print(len(ELEMENTS))
```

```python
def vectorize_composition(composition, elements):
    """ Converts and element composition dictionary to a vector. """
    total_n = sum(composition.values())
    vec = np.zeros(len(elements))
    for elem, n in composition.items():
        vec[elements.index(elem)] = n/total_n

    return vec
```

```python
def parse_data_vector(row):
    """ Parses a row in the superconductor DataFrame """

    # parse the composition dictionary
    composition_dict = ast.literal_eval(row['Composition'])

    # bulid the feature vector
    x_vector = np.concatenate([
        vectorize_composition(composition_dict, ELEMENTS),
        np.array([ row['Pressure (GPa)'] ]),
        np.array([ row['Tc (K)'] ])
    ])

    return x_vector
    
```

```python
data_x = np.array([
    parse_data_vector(row)
    for _, row in data_df.iterrows()
])
```

```python
print(data_x.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
data_z = scaler.fit_transform(data_x)

composition_z = data_z[:, :-2]

pressure_z = data_z[:, -2]
tc_z = data_z[:,-1]
```

```python
from sklearn.neighbors import KernelDensity

Tc_data = data_x[:,-1]
Tc_kde = KernelDensity(kernel='gaussian')
Tc_kde.fit(Tc_data.reshape(-1,1))
```

```python
T_eval_pts = np.linspace(0, np.max(Tc_data), 1000)

Tc_density = np.exp(Tc_kde.score_samples(T_eval_pts.reshape(-1, 1)))
Tc_density /= np.trapezoid(Tc_density, T_eval_pts)
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.grid()
plt.fill_between(T_eval_pts, Tc_density)
plt.xlabel('Temperature (K)')
plt.ylabel('Probability Density')
plt.show()
```
